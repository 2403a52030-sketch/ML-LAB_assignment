```python
# Import required libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
import pandas as pd

# Load the Excel file into a pandas DataFrame
df = pd.read_excel('/content/diabetes.xlsx')

# Display the first 5 rows of the DataFrame][P]
display(df.head())
```

```python
# Features (X) and Target (y)
X = df.drop('Outcome', axis=1)
y = df['Outcome']
```

```python
# Feature Scaling using StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Split dataset into training and testing sets (80:20)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
class LogisticRegressionGD:
    def __init__(self, learning_rate=0.01, epochs=1000):
        self.lr = learning_rate
        self.epochs = epochs

    # Sigmoid function
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    # Training using Gradient Descent
    def fit(self, X, y):
        self.m, self.n = X.shape

        # Initialize weights and bias
        self.weights = np.zeros(self.n)
        self.bias = 0

        # Gradient Descent loop
        for _ in range(self.epochs):
            linear_model = np.dot(X, self.weights) + self.bias
            y_pred = self.sigmoid(linear_model)

            # Compute gradients
            dw = (1 / self.m) * np.dot(X.T, (y_pred - y))
            db = (1 / self.m) * np.sum(y_pred - y)

            # Update parameters
            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    # Prediction function
    def predict(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        y_pred = self.sigmoid(linear_model)
        return [1 if i > 0.5 else 0 for i in y_pred]
```

```python
# Learning rates to test
learning_rates = [0.01, 0.001]

results = {}

for lr in learning_rates:
    print(f"\nTraining with Learning Rate = {lr}")

    model = LogisticRegressionGD(learning_rate=lr, epochs=1000)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    # Evaluation metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)

    results[lr] = (acc, prec, rec)

    print(f"Accuracy : {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall   : {rec:.4f}")
```

```python
print("\nFinal Comparison:")
print("Learning Rate | Accuracy | Precision | Recall")

for lr, (acc, prec, rec) in results.items():
    print(f"{lr:<14} | {acc:.4f}   | {prec:.4f}   | {rec:.4f}")
```
